Wap to find the sum of a first n number (using while)

```python
n = 5
sum = 0
for i in range(1,n+1) :
    sum += i
    print( "total sum = ", sum)
```

```python
n = 5
sum = 0
i = 1
while i<=n:
    sum += 1
    i += i
    print("total sum = ", sum)
```

How to remove duplicates from a list?

```python
#Method 1:
Number = [1,2,3,4,5,5,2,2,1,1,1,1]
result = list(set(Number))
print(result)
```

```python
#Method 2 (preserve order):
Number = [1,2,3,4,5,5,2,2,1,1,1,1]
result = []
for i in Number:
    if i not in result:
        result.append(i)
print(result)
```

*ARGS

Used for multiple positional arguments.

```python
def total(*args):
   return sum(args)
    
print(total(1,2,3,4,5))
```

**Kwargs

Used for multiple keyword arguments.

```python
def Student(**kwargs):
     print(kwargs)
Student(name = "Ali", Age = "22")
```

 Reverse a string

```python
S = "python"
reverse = S[ ::-1]
print(reverse)
```

Check palindrome string

```python
s = "madam"
if s[ :: -1]:
  print("Palindrome")
else:
  print("Not Palindrome")
```

Find maximum number in list

```python
number = [12,32,45,22,20,21]
maximum = number[0]
for i in number:
    if i>maximum:
        maximum = i 
print(maximum)
    
```
